## 2. The question: decision, action, cost of a wrong call

**Research Question**

Can observable search and content signals be used to rank pages that should be reviewed first for content refresh?

**Decision**

Which pages should receive a content review before others?

**Unit of Analysis**

One content page.

**Output**

A ranked list of pages with a refresh priority score, confidence score, and reason codes explaining why each page was recommended.

**Action**

A content editor or SEO specialist reviews the highest-ranked pages first and decides whether to update, expand, rewrite, monitor, or leave the content unchanged.

**Cost of a Wrong Recommendation**

A false positive wastes editorial effort by recommending pages that do not actually need attention.

A false negative is more costly because an important declining page may continue losing traffic before anyone notices.

Machine learning can help because multiple observable signals such as impressions, CTR, average position, freshness, engagement, and content age may interact in ways that are difficult to capture using a simple hand-written rule.

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains thousands of content pages, providing enough examples for exploratory analysis.

The proportion of declining pages shows that identifying pages requiring review is a meaningful ranking problem.

The weak correlation between search volume and impressions suggests that keyword popularity alone is not sufficient for deciding which pages deserve attention.

The average CTR decreases across lower ranking positions, indicating that ranking position is an important observable signal that may contribute to refresh decisions.

These observations suggest that combining multiple observable signals is likely to produce better recommendations than relying on any single metric.

```python

# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
else:
    # find the repo root from wherever this kernel started
    while not os.path.isdir("data/raw") and os.getcwd() != "/":
        os.chdir("..")

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found — are you at the repo root?"
print("Starter data found. You're ready.")


import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total Pages:", len(df))

declining_rate = (
    df["trend_direction"]
      .str.lower()
      .eq("down")
      .mean()
)

print(f"Declining Pages: {declining_rate:.2%}")

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between Search Volume and Impressions: {corr:.3f}")

visible = df[df["impressions_90d"] >= 100]

ctr_by_pos = (
    visible.groupby("position_tier")["ctr"]
           .mean()
           .sort_values(ascending=False)
)

print("\nAverage CTR by Position Tier")
print(ctr_by_pos.round(4))
```
